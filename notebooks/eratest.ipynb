```python
import cdsapi

dataset = "reanalysis-era5-single-levels"
request = {
    "product_type": ["reanalysis"],
    "variable": [
        "10m_u_component_of_wind",
        "10m_v_component_of_wind",
        "sea_surface_temperature"
    ],
    "year": ["2019"],
    "month": ["12"],
    "day": ["01"],
    "time": ["03:00"],
    "data_format": "netcdf",              # CHANGED from "grib"
    "download_format": "unarchived",
    "area": [-5.75, 106.75, -6, 107]
}

client = cdsapi.Client()
client.retrieve(dataset, request).download("era5_data.nc") # CHANGED to .nc
print("Download complete!")
```

```python
import xarray as xr
import numpy as np

# Open the NetCDF file (xarray will automatically use the installed netcdf4 engine)
ds = xr.open_dataset("era5_data.nc")

# Calculate total wind speed
ds['wind_speed'] = np.sqrt(ds.u10**2 + ds.v10**2)

# Convert to pandas DataFrame and drop NaNs
df = ds.to_dataframe().reset_index().dropna()

# Display the tabular data
display(df[['latitude', 'longitude', 'u10', 'v10', 'wind_speed', 'sst']])
```
